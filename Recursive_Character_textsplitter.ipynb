Text splitting from Documents

```python
#reading a PDF files
from langchain_community.document_loaders import PyPDFLoader
loader=PyPDFLoader('attention.pdf')
docs=loader.load()
```

```python
docs
```

HOw to recrusively split text by character

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter=RecursiveCharacterTextSplitter(chunk_size=5, chunk_overlap=2)
final_doc=text_splitter.split_documents(docs)
print(final_doc)
```

```python
print(final_doc[0])
print(final_doc[1])
```

Reading a txt file (speech.txt)

```python
from langchain_community.document_loaders import TextLoader

loader=TextLoader('speech.txt')
docs=loader.load()
docs
```

```python
speech=''
with open("speech.txt") as file:
    speech=file.read()


speech
```

```python
text_splitter=RecursiveCharacterTextSplitter(chunk_size=100,chunk_overlap=10)

text=text_splitter.create_documents([speech])
print(text)
# len(text)
```
