```python
# # using OpenAI
# from langchain_openai import ChatOpenAI
# llm=ChatOpenAI(model='gpt-4o')
# print(llm)

from transformers import pipeline
from langchain_huggingface import HuggingFacePipeline

#setup the pipeline
raw_pipe = pipeline(
    'text-generation',
    model='Qwen/Qwen2.5-0.5B-Instruct',
    max_new_tokens=150
)

# # Define our conversation
# messages=[
#     {'role':'system','content':'you are a technical assistant specializing in Machine Learning'},
#     {'role':'user','content':'I have two sentence with a cosine similarity of 0.9.What does this tell me about their relationship?'}
# ]

# #generating the response
# result=generator(messages,max_new_tokens=200)

# wrapping it for langchain
llm=HuggingFacePipeline(pipeline=raw_pipe)

# using invoke() to trigger langsmith trace instead of calling pipe directly
response=llm.invoke("what is feeling awesome?")
```
