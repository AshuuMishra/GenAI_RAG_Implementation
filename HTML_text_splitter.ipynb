HTMLHeaderTextSplitter is a "structure-aware" chunker that splits text at the HTML element level and adds
metadata for each header "relevant" to any given chunk. It can return chunks element by element or combir
elements with the same metadata, with the objectives of (a) keeping related text grouped (more or less)
semantically and (b) preserving context-rich information encoded in document structures. It can be used witn
other text splitters as part of a chunking pipeline.

>

```python
from langchain_text_splitter import HTMLHeaderTextSplitter

html_string="""

<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <title>HTML Splitter Test</title>
</head>
<body>

<h1>Understanding Attention Mechanisms</h1>

<p>
Attention mechanisms are a core component of modern transformer-based
models. They allow the model to focus on the most relevant parts of the
input sequence when generating outputs.
</p>

<h2>Why Attention Matters</h2>

<p>
Traditional sequence models struggled with long-range dependencies.
Attention solves this by computing weighted relationships between
tokens, regardless of their distance in the sequence.
</p>

<h3>Key Benefits</h3>

<ul>
    <li>Improved context handling</li>
    <li>Parallel computation</li>
    <li>Better long-text understanding</li>
</ul>

<h2>Self-Attention Explained</h2>

<p>
Self-attention works by projecting input tokens into query, key, and
value vectors. The similarity between queries and keys determines how
information flows through the network.
</p>

<h2>Conclusion</h2>

<p>
Attention is not just a mechanism — it is the reason large language
models scale so effectively. Understanding it unlocks deeper insight
into modern AI systems.
</p>

</body>
</html>

"""
headers_to_split_on=[
    ("h1","header 1")
    ("h2","header 2")
    ("h3","header 3")
]
html_splitter=HTMLHeaderTextSplitter(headers_to_split_on)
html_header_splits=html_splitter.split_text(html_string)
html_header_splits
```
