Character Text Splitter

This is the simples method .This splits based on a given character sequence which defautls to "\n\n". Chucnk length is measured by number of characters.

1.Text is split by single character separator

2.and measured by number of characters

```python
from langchain_community.document_loaders import TextLoader
loader=TextLoader('speech.txt')
docs=loader.load()
docs
```

```python
from langchain_text_splitters import CharacterTextSplitter
text_splitter=CharacterTextSplitter(separator="\n\n",chunk_size=100,chunk_overlap=20)
final=text_splitter.split_documents(docs)
# final
```
