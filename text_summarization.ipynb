1. Manually create messages

2. Create a normal Python template

       ↓
3. Turn the template into a LangChain PromptTemplate

4. Create a ChatPromptTemplate and chain it directly to the LLM

```python
from langchain_core.prompts import PromptTemplate
# from re import TEMPLATE
from langchain_core.messages import HumanMessage,AIMessage,SystemMessage
from langchain_core.prompts import ChatPromptTemplate

chat_message = [
    SystemMessage(content="you;re an expert in summarising the content"),
    HumanMessage(content=f"Please provide the content to be summarise\n ,Text :{summary}")
]

response = llm.invoke(chat_message)
response

# -----------------------------------------------------------------------------
generic_template = """
Write a summary of the following speech 
Speech : {input}
enter the language you to summary : {language}
"""

prompt=generic_template.format(input=summary,language ="hindi")
response = llm.invoke(prompt)
response

prompt = PromptTemplate(
    input_variables=['input',"language"],
    template=generic_template
)
response = prompt.format(input=summary,language ="hindi")
response

# ----------------------------------------------------------------------
# PROMPT TEMPLATE TEXT SUMMARIZATION

prompt = ChatPromptTemplate.from_messages([
    ("system","you're an expert summariser , summarise the data"),
    ("human","provide the data to be summarise {input}")
])

chain=prompt | llm
response = chain.invoke({"input":summary})
response
```

## Stuff Document Chain Summarisation

```python
from langchain_community.document_loaders import PyPDFLoader
# from langchain_community import __l
documents = PyPDFLoader(r"C:\Users\CloudJournee\Desktop\python\python\langchain_update\GenAI_RAG_Implementation\Amazon Sagemaker.pdf")
docs = documents.load_and_split()
docs
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_classic.chains.summarize import load_summarize_chain   

template = """"
write a concise and short summary of privded pdf'.
Document :{text}
"""
prompt =PromptTemplate(
    input_variables=['text'],
    template=template
)


chain = load_summarize_chain(llm, chain_type="stuff",prompt=prompt,verbose=True)
output_summary = chain.run(docs)
output_summary
```

## similary we use map reduce for text summarisation 

```python
from h11._abnf import chunk_size
from langchain_core.prompts import PromptTemplate
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

corpus = PyPDFLoader(r"C:\Users\CloudJournee\Desktop\python\python\langchain_update\GenAI_RAG_Implementation\Amazon Sagemaker.pdf")
docs = corpus.load()
final_docs=text_splitter=RecursiveCharacterTextSplitter(chunk_size=2000,chunk_overlap=200).split_documents(docs)
final_docs
```

```python
len(final_docs)
```

```python
from langchain_classic.chains.summarize import load_summarize_chain

template ="""
assign you as expert summarise in the history ever has 
summary : {text}.
"""

map_prompt_template = PromptTemplate(
    input_variables=['text'],
    template=template
)


```

```python
final_prompt ="""
provide the final summary of the entire speech with these important points.
add a motivatinal title , start the precise summary with and introduction and summarise in the points for the data
content : {text}
"""

final_prompt_template =PromptTemplate(
    input_variables=["text"],
    template=final_prompt
)
final_prompt_template
```

```python
summary_chain=load_summarize_chain(
    llm=llm,
    chain_type="map_reduce",
    verbose=True,
    map_prompt=map_prompt_template ,
    combine_prompt =final_prompt_template)

    
summary_response= summary_chain.invoke(final_docs)
summary_response
```
