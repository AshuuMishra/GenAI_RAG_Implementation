4. Create a ChatPromptTemplate and chain it directly to the LLM

```python
from langchain_core.prompts import PromptTemplate
# from re import TEMPLATE
from langchain_core.messages import HumanMessage,AIMessage,SystemMessage
from langchain_core.prompts import ChatPromptTemplate

chat_message = [
    SystemMessage(content="you;re an expert in summarising the content"),
    HumanMessage(content=f"Please provide the content to be summarise\n ,Text :{summary}")
]

response = llm.invoke(chat_message)
response

# -----------------------------------------------------------------------------
generic_template = """
Write a summary of the following speech 
Speech : {input}
enter the language you to summary : {language}
"""

prompt=generic_template.format(input=summary,language ="hindi")
response = llm.invoke(prompt)
response

prompt = PromptTemplate(
    input_variables=['input',"language"],
    template=generic_template
)
response = prompt.format(input=summary,language ="hindi")
response

# ----------------------------------------------------------------------
# PROMPT TEMPLATE TEXT SUMMARIZATION

prompt = ChatPromptTemplate.from_messages([
    ("system","you're an expert summariser , summarise the data"),
    ("human","provide the data to be summarise {input}")
])

chain=prompt | llm
response = chain.invoke({"input":summary})
response
```
