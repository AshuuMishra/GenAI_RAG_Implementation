***Streaming***

Most models can stream their output content while it is being generated.By displaying output progressively , streaming significantly imporves user experience ,
particularly for loonger responses.Calling stream() returns an iterators that yield output chunks as they are produced. you can use a loop to process each chunk in real time .

````python
# model.invoke("translate the word - beshram into english ")

# now will try with stream()

model.stream("translate the word - beshram into english ")

for chunk in model.stream(" translate the word - beshram into english"):
    print(chunk)
````

**Batch**

Batching a collection of independent requests to a model can significantly improve performance and reduce costs ,as the processing can be done in paralle.

````python
response=model.batch([
    "According to Indain Astronomy , list down the days of week along with color of dress be wore on that day .",
    "Why do we feel sleepy after having food is there any reason or its just assumption?",
    "what happens if we breach company law or bond of committment ..?"
])

for ans in response:
    print(ans)
````

````python
response=model.batch([
    "According to Indain Astronomy , list down the days of week along with color of dress be wore on that day .",
    "Why do we feel sleepy after having food is there any reason or its just assumption?",
    "what happens if we breach company law or bond of committment ..?"],
    config={
        "max_concurrency":2,  # limit to 2 parallel calls 
    }
)

for ans in response:
    # print(ans)
    print(ans.content)
````

## Understanding `batch()` and `max_concurrency`

The `batch()` method is used to process multiple inputs efficiently by executing requests in parallel instead of one after another.

### Without Batch Processing

```python
for question in questions:
    model.invoke(question)
```

Execution is sequential:

```
Q1 → wait
Q2 → wait
Q3 → wait
...
```

If each request takes 2 seconds and there are 100 questions:

```
Total Time = 100 × 2 = 200 seconds
```

---

### With Batch Processing

```python
model.batch(
    questions,
    config={"max_concurrency": 10}
)
```

LangChain creates up to 10 concurrent workers.

Execution:

```
Worker1 → Q1
Worker2 → Q2
...
Worker10 → Q10
```

As soon as a worker finishes, it picks the next question from the queue.

```
Worker1 finishes Q1 → starts Q11
Worker2 finishes Q2 → starts Q12
...
```

---

### Meaning of `max_concurrency`

```python
config={"max_concurrency": 2}
```

means:

- Maximum 2 requests can run simultaneously.
- It does **not** split 100 questions into two groups of 50.
- It creates 2 workers that continuously process questions until all inputs are completed.

Visualization:

```
Queue:
Q1 Q2 Q3 Q4 Q5 ...

Worker1 → Q1 → Q3 → Q5 ...
Worker2 → Q2 → Q4 → Q6 ...
```

---

### Why Use It?

1. Faster processing of large datasets.
2. Better utilization of API throughput.
3. Useful for:
   - Document summarization
   - Sentiment analysis
   - LLM evaluation
   - Bulk content generation
   - Embedding generation

---

### Trade-off

Higher concurrency = Faster execution

But too much concurrency may cause:

- API rate limit errors (`429`)
- Increased memory usage
- Provider throttling

Choose a value that balances speed and API limits.
